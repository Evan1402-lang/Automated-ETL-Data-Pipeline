```python
import pandas as pd
import sqlite3

def run_etl():
    # 1. EXTRACT: Load raw messy data
    raw_df = pd.read_csv('raw_orders.csv')

    # 2. TRANSFORM: Data Cleaning & Standardizing
    raw_df['order_date'] = pd.to_datetime(raw_df['order_date'], errors='coerce')
    raw_df = raw_df.dropna(subset=['order_id', 'order_date'])

    # Text normalization
    raw_df['customer_city'] = raw_df['customer_city'].str.title().str.strip()
    raw_df['amount'] = raw_df['amount'].abs()  # Fix negative transaction errors

    # 3. LOAD: Save into SQLite Database
    conn = sqlite3.connect('warehouse.db')
    raw_df.to_sql('fact_orders', conn, if_exists='replace', index=False)
    conn.close()
    print("ETL Pipeline completed successfully! Data loaded into 'fact_orders'.")

if __name__ == '__main__':
    run_etl()
```
